## 1. Linked List

```python
class Node:
  def __init__(self,data):
    self.data = data
    self.next = None
    head = None
    tail = None

    newnode = Node(10)

    if head ==None and tail == None :
      head =newnode
      tail = newnode

    newnode = Node(20)
    tail.next = newnode
    tail = newnode

    temp = head
    while temp != None:
      print(temp.data)
      temp = temp.next
    print(None)


```

## 3. Stack

```python
stack = []
top = -1
def push(data):
  global top
  stack.append(data)
  top += 1
  print(data)

def pop():
  global top
  if top ==-1:
    print("Stack is Empty")
  else:
    print("Delect Element",stack[top])
    stack.pop()
    top -= 1

def peek():
  global top
  if top == -1:
    print("stack is empty")
  else:
    print("stack top:",stack[top])

push(10)
push(20)
push(30)
push()
peek()
```
