```python
#1. Count frequency of each element in an array
def count_frequency(nums):
  freq = {}
  for num in nums:
    if num in freq:
      freq[num] += 1
    else:
      freq[num] =1
  for num , count in freq.items():
    print(num ,":",count)



  #  count[num] = count.get(num,0) + 1

count_frequency([1,2,1,6,8,9,4,5,8])
```

```python
#2. Find all the repeating elements
def All_repeated(nums):
     freq = {}
     for num in nums:
      if num in freq:
        freq[num] += 1
      else:
        freq[num] =1
     for i in freq:
       if freq[i]>1:
        print(i)
All_repeated([5,6,8,6,4,9,1,2,3,4,1])
```

```python
#3. Find all the non repeating elements
def All_nonrepeated(nums):
     freq = {}
     for num in nums:
      if num in freq:
        freq[num] += 1
      else:
        freq[num] =1
     for i in freq:
       if freq[i]==1:
        print(i)
All_nonrepeated([5,6,8,6,4,9,1,2,3,4,1])
```

```python
#4. Find Distinct elements
def distinct(nums):
     freq = {}
     for num in nums:
      if num in freq:
        freq[num] += 1
      else:
        freq[num] =1
     for i in freq:
       print(i)
distinct([5,6,8,6,4,9,1,2,3,4,1])
```

```python
#7. First Repeating Element
def first_repeated(nums):
    freq = {}

    for num in nums:
        if num in freq:
            break
        else:
            freq[num] = 1

    return num

print(first_repeated([5,6,8,6,4,9,1,2,3,4,1]))
```

```python
#8. Non-Repeating Element
def non_repeated(nums):
    freq = {}

    for num in nums:
        if num in freq:
           freq[num] += 1
        else:
            freq[num] = 1

    for num in freq:
      if freq[num] ==1:
        break
    return num
non_repeated([5,6,8,6,4,5,9,1,2,3,4,1])
```
